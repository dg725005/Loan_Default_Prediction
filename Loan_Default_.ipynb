# Importing essential libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns",None)
```

# Loading CSV file

```python
df = pd.read_csv(r"E:\Loan_Default_Dataset\2\Loan_default.csv")
```

```python
y = df.Default

frame = pd.get_dummies(df,columns=['Education', 'EmploymentType', 'MaritalStatus',
       'HasMortgage', 'HasDependents', 'LoanPurpose', 'HasCoSigner'])
frame.replace({True:1.0, False:0.0},inplace=True)
frame = frame.drop(columns=['Default'])
frame.info()
```

# Treating imbalance of the dataset using SMOTE

```python
X = frame.iloc[:,1:]
X.info()
```

```python
y.info()
```

```python
# pip install imbalanced-learn

from imblearn.over_sampling import SMOTE
smote = SMOTE(sampling_strategy='minority')
X_res, y_res = smote.fit_resample(X, y)
```

# Clean-up

```python
#df,y,X,frame = [], [], [], []
```

```python
print("Dataframe size =", len(y_res))

default_percent = ((len(y_res[y_res==1.0]))/len(y_res))*100.00

print("Percentage of default cases =",round(default_percent,2),"%")
```

```python
tem = []
for r in range(len(y_res)):
    if y_res[r] == 0.0:
        tem.append([0.0,1.0])
    else:
        tem.append([1.0,0.0])    
```

# Train-Test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_res,tem,random_state=38,test_size=0.10)
```

# Standardization of training data

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.models import Sequential

model = Sequential()

model.add(Dense(256, activation='relu', input_dim = X_train.shape[1]))
model.add(Dropout(0.20))
model.add(Dense(256+128, activation='relu'))
model.add(Dropout(0.20))
model.add(Dense(2, activation = 'softmax'))

model.summary()
```

```python
model.compile(loss='categorical_crossentropy',metrics=['accuracy'],optimizer='adam')
```

```python
model.fit(np.array(X_train), np.array(y_train), epochs=10, batch_size=64, verbose=1,validation_data=(np.array(X_test), np.array(y_test)))
```

# Using Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

classifier = RandomForestClassifier(n_estimators=196,max_depth=64,random_state=49)

classifier.fit(X_train,y_train)

```

# Performance Metrices

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score

print("Accuracy Score = " + str(accuracy_score(y_test, classifier.predict(X_test))*100.00))

print("Precision Score = " + str(precision_score(y_test, classifier.predict(X_test),average='weighted')*100.00))

print("Recall Score = " + str(recall_score(y_test, classifier.predict(X_test),average='weighted')*100.00))

print("F1 Score = " + str(f1_score(y_test, classifier.predict(X_test),average='weighted')*100.00))

print("ROC-AUC Score = " + str(roc_auc_score(y_test, classifier.predict(X_test),average='weighted')*100.00))
```
